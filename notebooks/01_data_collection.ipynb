```python
import serpapi
import os
import pandas
from datetime import datetime
from dotenv import load_dotenv

load_dotenv('../src/.env')
SERPAPI_KEY = os.getenv('SERPAPI_KEY')
PRODUCT_ID = os.getenv('PRODUCT_ID')

client = serpapi.Client(api_key=SERPAPI_KEY)
```

```python
dfs = []
next_page = ''
i = 0

results = client.search(
            engine='google_play_product',
            product_id=PRODUCT_ID,
            store='apps',
            all_rewiews=True,
            num=199,
            platform='phone',
            gl='br',
            hl='pt-br',
)

for page in results.yield_pages():
    dfs.append(
        pandas.DataFrame(
            page['reviews']
        )
    )
    i+=1
    if i>1103: break

date = str(datetime.now().day).zfill(2) + '_' + str(datetime.now().month).zfill(2) + '_' + str(datetime.now().year)

data = pandas.concat(dfs)
data.to_parquet(f'../data/raw/{PRODUCT_ID}_{date}_BR.parquet')

print("Total reviews: ", data['snippet'].nunique())
```
